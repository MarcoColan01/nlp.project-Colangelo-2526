```python
from pathlib import Path
PROJECT_ROOT = Path('..').resolve()
RAW_DIR = PROJECT_ROOT / 'data' / 'raw'
PROCESSED_DIR = PROJECT_ROOT / 'data' / 'processed'
RAW_DIR.mkdir(parents=True, exist_ok=True)
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

kaggle_csv = RAW_DIR / 'Connections_data.csv'
kaggle_csv 
```

```python
import pandas as pd 
df = pd.read_csv(kaggle_csv)
print('Loaded: ', df.shape)
display(df.head())
```

```python
import sys 
sys.path.append(str((PROJECT_ROOT / 'src').resolve()))
from data.kaggle_to_huggingface import kaggle_rows_to_puzzles, KaggleToHFConfig, save_jsonl, quick_validate
```

```python
config = KaggleToHFConfig(seed=1234, shuffle_words = True, min_date_inclusive = '2025-03-25')
puzzles = kaggle_rows_to_puzzles(df, config=config)
n, suspect = quick_validate(puzzles)
print('Puzzles: ', n, '| Suspect: ', suspect)
puzzles[0]
```

```python
out_jsonl = PROCESSED_DIR / 'kaggle_hf_like.jsonl'
save_jsonl(puzzles, out_jsonl)
print('Saved:', out_jsonl)
print('Bytes:', out_jsonl.stat().st_size)
```
