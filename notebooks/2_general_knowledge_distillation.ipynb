# Fase 1: General Distillation (TinyBERT Protocol adapted to Llama)

In this phase, we use the original Teacher (DeepSeek-R1-Distill-Llama) and an unsupervised corpus (SWOW dataset) for the general knowledge distillation on the student (Llama-3.2-Instruct-1B)

**What will be distilled from the teacher to the student**
- Embedding Layer
- Transformer Hidden States
- Attention Matrices 
- NO Prediction Layer (logits) distilation for this phase

```python
import sys 
import os 
from pathlib import Path 
current_dir = Path(os.getcwd())
project_root = current_dir.parent 
if str(project_root) not in sys.path:
    sys.path.append(str(project_root))
```

```python
from src.teacher_training import GeneralDistillation

data_dir = project_root / "data" / "processed"
train_file = data_dir / "swow_train.jsonl"
output_dir = project_root / "checkpoints" / "student_phase1_general"
```

```python
TEACHER_ID = "deepseek-ai/DeepSeek-R1-Distill-Llama-8B"
STUDENT_ID = "meta-llama/Llama-3.2-1B-Instruct"
distiller = GeneralDistillation(
    student_model_id=STUDENT_ID,
    teacher_model_id=TEACHER_ID,
    output_dir=str(output_dir)
    )
print(distiller.teacher.hf_device_map)
```

```python
distiller.train(
    train_file=str(train_file),
    epochs=1,
    batch_size=4,
    accumulation_steps=4
)
```
